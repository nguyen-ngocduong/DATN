## Cell 2: Load Data with Optimized Dtypes

**Note:** For large datasets, we'll use optimized dtypes to reduce RAM usage

```python
# Google Drive đã được gắn kết (mounted) và điều hướng tự động ở Cell 1.
if IN_COLAB:
    print(f"Working directory hiện tại: {os.getcwd()}")
else:
    print("Môi trường Local: sẵn sàng nạp dữ liệu.")
```

```python
# ==============================================================================
# Cell 2: Nạp Dữ Liệu với Kiểu Dữ Liệu Tối Ưu (Memory-Optimized Dtypes)
# ==============================================================================
import os

# Define optimized dtypes for memory efficiency
dtype_dict = {
    "flow_duration": "float32",
    "Header_Length": "float32",
    "Protocol Type": "float32",
    "Duration": "float32",
    "Rate": "float32",
    "Srate": "float32",
    "Drate": "float32",
    "fin_flag_number": "float32",
    "syn_flag_number": "float32",
    "rst_flag_number": "float32",
    "psh_flag_number": "float32",
    "ack_flag_number": "float32",
    "ece_flag_number": "float32",
    "cwr_flag_number": "float32",
    "ack_count": "float32",
    "syn_count": "float32",
    "fin_count": "float32",
    "urg_count": "float32",
    "rst_count": "float32",
    "HTTP": "float32",
    "HTTPS": "float32",
    "DNS": "float32",
    "Telnet": "float32",
    "SMTP": "float32",
    "SSH": "float32",
    "IRC": "float32",
    "TCP": "float32",
    "UDP": "float32",
    "DHCP": "float32",
    "ARP": "float32",
    "ICMP": "float32",
    "IPv": "float32",
    "LLC": "float32",
    "Tot sum": "float32",
    "Min": "float32",
    "Max": "float32",
    "AVG": "float32",
    "Std": "float32",
    "Tot size": "float32",
    "IAT": "float32",
    "Number": "float32",
    "Magnitue": "float32",
    "Radius": "float32",
    "Covariance": "float32",
    "Variance": "float32",
    "Weight": "float32",
    "label": "category"
}

# Load data
print("Loading data with optimized dtypes...")

# Tìm kiếm file train.csv tự động
candidates = [
    "train.csv",
    "/content/drive/MyDrive/do_an/train.csv",
    "data/train.csv",
    "plots/train.csv",
    "../train.csv"
]
data_path = next((p for p in candidates if os.path.exists(p)), None)
if not data_path:
    raise FileNotFoundError(f"Không tìm thấy train.csv trong các đường dẫn: {candidates}")

print(f"Đang đọc dữ liệu từ: {data_path}...")
df = pd.read_csv(data_path, dtype=dtype_dict)

print("Data loaded successfully!")
print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
```

## Cell 3: Basic Data Info

```python
# Basic data information
print('='*60)
print('BASIC DATA INFORMATION')
print('='*60)

print(f'\nDataset Shape: {df.shape[0]:,} rows x {df.shape[1]} columns')

# Memory usage
memory_usage_mb = df.memory_usage(deep=True).sum() / 1024**2
print(f'Memory Usage: {memory_usage_mb:.2f} MB')

# Column names
print(f'\nColumn Names:')
for i, col in enumerate(df.columns, 1):
    print(f'   {i:2d}. {col}')
```

## Cell 4: Data Types

```python
# Check data types
print('='*60)
print('DATA TYPES')
print('='*60)

print(df.dtypes)
```

## Cell 5: Missing Values Analysis

```python
# Missing values analysis
print('='*60)
print('MISSING VALUES ANALYSIS')
print('='*60)

missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
})

# Filter only columns with missing values
missing_df = missing_df[missing_df['Missing Count'] > 0]

if len(missing_df) > 0:
    print(f'\nFound {len(missing_df)} columns with missing values:')
    print(missing_df.sort_values('Missing %', ascending=False))
else:
    print('\nNo missing values found!')

# Also check for infinite values
print('\n\nChecking for infinite values...')
inf_cols = []
for col in df.select_dtypes(include=['float32']).columns:
    inf_count = np.isinf(df[col]).sum()
    if inf_count > 0:
        inf_cols.append((col, inf_count))

if inf_cols:
    print(f'\nFound columns with infinite values:')
    for col, count in inf_cols:
        print(f'   - {col}: {count:,} ({count/len(df)*100:.2f}%)')
else:
    print('\nNo infinite values found!')
```

## Cell 6: Duplicate Rows Check

```python
# Duplicate rows analysis
print('='*60)
print('DUPLICATE ROWS ANALYSIS')
print('='*60)

duplicates = df.duplicated().sum()
print(f'\nTotal duplicate rows: {duplicates:,}')
print(f'   Percentage: {duplicates/len(df)*100:.2f}%')

if duplicates > 0:
    print(f'\nConsider removing duplicates for cleaner analysis')
else:
    print('\nNo duplicate rows found!')
```

## Cell 7: Statistical Summary

```python
# Statistical summary
print('='*60)
print('STATISTICAL SUMMARY')
print('='*60)

desc = df.describe().T
print(desc)
```

## Cell 8: Target Variable Distribution

```python
# Target variable (label) distribution
print('='*60)
print('TARGET VARIABLE DISTRIBUTION')
print('='*60)

label_counts = df['label'].value_counts()
label_pct = df['label'].value_counts(normalize=True) * 100

# Create summary dataframe
label_summary = pd.DataFrame({
    'Count': label_counts,
    'Percentage (%)': label_pct.round(2)
})

print(f'\nTotal unique labels: {len(label_counts)}')
print(f'\n{label_summary}')
```

## Cell 9: Visualize Class Distribution

```python
# Visualize class distribution
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# Plot 1: Bar chart of class distribution
ax1 = axes[0]
colors = plt.cm.Set3(np.linspace(0, 1, len(label_counts)))
bars = ax1.barh(range(len(label_counts)), label_counts.values, color=colors)
ax1.set_yticks(range(len(label_counts)))
ax1.set_yticklabels(label_counts.index, fontsize=9)
ax1.set_xlabel('Count', fontsize=11)
ax1.set_title('Attack Type Distribution (Count)', fontsize=13, fontweight='bold')
ax1.invert_yaxis()

# Add count labels on bars
for i, (bar, count) in enumerate(zip(bars, label_counts.values)):
    ax1.text(count + label_counts.max()*0.01, i, f'{count:,}', va='center', fontsize=8)

# Plot 2: Pie chart
ax2 = axes[1]
top_10 = label_counts.head(10)
other = label_counts.iloc[10:].sum()
if other > 0:
    pie_data = pd.concat([top_10, pd.Series({'Other': other})])
else:
    pie_data = top_10

colors_pie = plt.cm.Set3(np.linspace(0, 1, len(pie_data)))
wedges, texts, autotexts = ax2.pie(pie_data.values, labels=pie_data.index, 
                                    autopct='%1.1f%%', colors=colors_pie,
                                    pctdistance=0.75, labeldistance=1.1)
ax2.set_title('Attack Type Distribution (Percentage)', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.savefig('plots/01_class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nPlot saved to: plots/01_class_distribution.png')
```

## Cell 10: Group Attacks by Category

```python
# Group attacks by category
print('='*60)
print('ATTACK CATEGORY ANALYSIS')
print('='*60)

# Define attack categories
attack_mapping = {
    # DDoS attacks
    'DDoS-SYN_Flood': 'DDoS',
    'DDoS-TCP_Flood': 'DDoS',
    'DDoS-ICMP_Flood': 'DDoS',
    'DDoS-UDP_Flood': 'DDoS',
    'DDoS-ACK_Fragmentation': 'DDoS',
    'DDoS-PSHACK_Flood': 'DDoS',
    'DDoS-RSTFINFlood': 'DDoS',
    'DDoS-SynonymousIP_Flood': 'DDoS',
    
    # DoS attacks
    'DoS-SYN_Flood': 'DoS',
    'DoS-TCP_Flood': 'DoS',
    'DoS-UDP_Flood': 'DoS',
    
    # Mirai attacks
    'Mirai-greeth_flood': 'Mirai',
    'Mirai-greip_flood': 'Mirai',
    'Mirai-udpplain': 'Mirai',
    
    # Reconnaissance
    'Recon-HostDiscovery': 'Recon',
    
    # Benign
    'BenignTraffic': 'Benign'
}

# Create category column
df['attack_category'] = df['label'].map(attack_mapping)

# Count by category
category_counts = df['attack_category'].value_counts()
category_pct = df['attack_category'].value_counts(normalize=True) * 100

category_summary = pd.DataFrame({
    'Count': category_counts,
    'Percentage (%)': category_pct.round(2)
})

print(f'\nAttack Categories:')
print(category_summary)

# Visualize
fig, ax = plt.subplots(figsize=(10, 6))
colors_cat = ['#e74c3c', '#e67e22', '#9b59b6', '#3498db', '#2ecc71']
bars = ax.bar(category_counts.index, category_counts.values, color=colors_cat[:len(category_counts)])

ax.set_xlabel('Attack Category', fontsize=11)
ax.set_ylabel('Count', fontsize=11)
ax.set_title('Attack Category Distribution', fontsize=14, fontweight='bold')

# Add labels
for bar, count, pct in zip(bars, category_counts.values, category_pct.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + category_counts.max()*0.01,
            f'{count:,}\n({pct:.1f}%)', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.savefig('plots/02_category_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nPlot saved to: plots/02_category_distribution.png')
```

## Cell 11: Class Imbalance Analysis

```python
# Class imbalance analysis
print('='*60)
print('CLASS IMBALANCE ANALYSIS')
print('='*60)

imbalance_ratio = label_counts.max() / label_counts.min()
print(f'\nImbalance Ratio (max/min): {imbalance_ratio:,.2f}:1')

# Identify minority classes (< 1%)
minority_classes = label_pct[label_pct < 1]
print(f'\nMinority Classes (< 1%): {len(minority_classes)}')
if len(minority_classes) > 0:
    print(minority_classes.sort_values())

# Identify majority classes (> 10%)
majority_classes = label_pct[label_pct > 10]
print(f'\nMajority Classes (> 10%): {len(majority_classes)}')
if len(majority_classes) > 0:
    print(majority_classes.sort_values(ascending=False))
```

## Cell 12: Feature Correlation Analysis

```python
# Feature correlation analysis (sampled for memory efficiency)
print('='*60)
print('FEATURE CORRELATION ANALYSIS')
print('='*60)

# Sample 100k rows for correlation analysis (memory efficient)
sample_size = 100000
df_sample = df.sample(n=min(sample_size, len(df)), random_state=42)
print(f'\nUsing {len(df_sample):,} samples for correlation analysis')

# Select only numeric columns
numeric_cols = df_sample.select_dtypes(include=['float32', 'float64', 'int32', 'int64']).columns
print(f'Number of numeric features: {len(numeric_cols)}')

# Calculate correlation matrix
corr_matrix = df_sample[numeric_cols].corr()

print('\nCorrelation matrix calculated successfully!')
```

## Cell 13: Visualize Correlation Heatmap

```python
# Visualize correlation heatmap
plt.figure(figsize=(16, 14))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, cmap='RdBu_r', center=0, 
            annot=False, square=True, linewidths=0.5,
            cbar_kws={'shrink': 0.8})
plt.title('Feature Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('plots/03_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nPlot saved to: plots/03_correlation_heatmap.png')
```

## Cell 14: Highly Correlated Features

```python
# Find highly correlated feature pairs
print('='*60)
print('HIGHLY CORRELATED FEATURES (|r| > 0.9)')
print('='*60)

# Get upper triangle of correlation matrix
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Find features with high correlation
high_corr = [(col, row, upper.loc[row, col]) 
             for col in upper.columns 
             for row in upper.index 
             if pd.notna(upper.loc[row, col]) and abs(upper.loc[row, col]) > 0.9]

if len(high_corr) > 0:
    print(f'\nFound {len(high_corr)} highly correlated pairs:')
    for f1, f2, corr in sorted(high_corr, key=lambda x: abs(x[2]), reverse=True)[:20]:
        print(f'   {f1} <-> {f2}: {corr:.3f}')
else:
    print('\nNo highly correlated features found!')
```

## Cell 15: Feature Distribution Analysis

```python
# Feature distribution analysis
print('='*60)
print('FEATURE DISTRIBUTION ANALYSIS')
print('='*60)

# Sample for visualization
sample_10k = df.sample(n=min(10000, len(df)), random_state=42)

# Key features to visualize
key_features = ['flow_duration', 'Header_Length', 'Rate', 'Srate', 'Drate', 
                'Tot size', 'IAT', 'Number', 'Magnitue', 'Radius']

# Filter existing features
key_features = [f for f in key_features if f in df.columns]

fig, axes = plt.subplots(3, 4, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(key_features[:12]):
    ax = axes[i]
    data = sample_10k[col].dropna()
    ax.hist(data, bins=50, color='steelblue', edgecolor='white', alpha=0.7)
    ax.set_title(f'{col}', fontsize=10, fontweight='bold')
    ax.set_xlabel('')

# Hide empty subplots
for i in range(len(key_features), len(axes)):
    axes[i].set_visible(False)

plt.suptitle('Feature Distributions (Sample of 10,000)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('plots/04_feature_distributions.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nPlot saved to: plots/04_feature_distributions.png')
```

## Cell 16: Protocol Distribution Analysis

```python
# Protocol distribution analysis
print('='*60)
print('PROTOCOL DISTRIBUTION')
print('='*60)

protocol_cols = ['HTTP', 'HTTPS', 'DNS', 'Telnet', 'SMTP', 'SSH', 'IRC', 'TCP', 'UDP', 'DHCP', 'ARP', 'ICMP', 'IPv', 'LLC']
protocol_cols = [c for c in protocol_cols if c in df.columns]

protocol_means = df[protocol_cols].mean().sort_values(ascending=False)

print('\nMean Protocol Usage:')
print(protocol_means.to_string())

# Visualize
fig, ax = plt.subplots(figsize=(12, 6))
colors = plt.cm.Blues(np.linspace(0.3, 0.9, len(protocol_means)))
bars = ax.bar(protocol_means.index, protocol_means.values, color=colors)

ax.set_xlabel('Protocol', fontsize=11)
ax.set_ylabel('Mean Value', fontsize=11)
ax.set_title('Protocol Usage Distribution', fontsize=14, fontweight='bold')
ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('plots/05_protocol_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nPlot saved to: plots/05_protocol_distribution.png')
```

## Cell 17: Packet Flag Distribution

```python
# Packet flag distribution analysis
print('='*60)
print('PACKET FLAG DISTRIBUTION')
print('='*60)

flag_cols = ['fin_flag_number', 'syn_flag_number', 'rst_flag_number', 
             'psh_flag_number', 'ack_flag_number', 'ece_flag_number', 'cwr_flag_number']
flag_cols = [c for c in flag_cols if c in df.columns]

# Calculate statistics for flags
flag_stats = pd.DataFrame({
    'Mean': df[flag_cols].mean(),
    'Max': df[flag_cols].max(),
    'Sum': df[flag_cols].sum(),
    'Non-Zero %': (df[flag_cols] > 0).mean() * 100
}).round(4)

print(flag_stats)
```

## Cell 18: Summary Statistics by Attack Category

```python
# Feature statistics by attack category
print('='*60)
print('FEATURE STATISTICS BY ATTACK CATEGORY')
print('='*60)

# Key features for comparison
compare_features = ['Rate', 'Srate', 'Drate', 'flow_duration', 'Header_Length']
compare_features = [f for f in compare_features if f in df.columns]

# Group by attack category
grouped = df.groupby('attack_category')[compare_features].agg(['mean', 'median', 'std'])

print('\nMean values by attack category:')
print(grouped['mean'].round(2))
```
